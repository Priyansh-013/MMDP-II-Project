## Extract Flag Images and Calculate Aspect Ratios

```python
import os
from PIL import Image

# List to store flag aspect ratios
aspect_ratios = []

# Path to the flags folder
flags_folder = '../A_DataCollection/flags_png'

# Check if the folder exists
if os.path.exists(flags_folder):
    # Iterate through all images in the flags_png folder
    for filename in os.listdir(flags_folder):
        if filename.endswith(".png"):
            img_path = os.path.join(flags_folder, filename)
            img = Image.open(img_path)
            width, height = img.size
            aspect_ratio = width / height
            aspect_ratios.append(aspect_ratio)
else:
    print("The 'flags_png' folder does not exist.")

# Print unique aspect ratios and their counts
unique_ratios = set(aspect_ratios)
ratio_counts = {ratio: aspect_ratios.count(ratio) for ratio in unique_ratios}
print(ratio_counts)
```

## Validate Common Aspect Ratios






```python
# Common aspect ratios
common_ratios = {
    '1:2': 2.0,
    '2:3': 1.5,
    '3:5': 1.6  # Approximation for 3:5
}

# Validate common ratios
for name, ratio in common_ratios.items():
    if ratio in unique_ratios:
        print(f"{name} is present with count: {ratio_counts[ratio]}")
    else:
        print(f"{name} is not present.")
```

## Identify Influences and Patterns

```python
# Group flags by aspect ratios
ratio_groups = {}
for ratio in unique_ratios:
    flags_with_ratio = [filename for filename, aspect_ratio in zip(os.listdir(flags_folder), aspect_ratios) if aspect_ratio == ratio]
    ratio_groups[ratio] = flags_with_ratio

# Print flags grouped by aspect ratios
for ratio, flags in ratio_groups.items():
    print(f"Aspect Ratio: {ratio}, Flags: {flags}")
```

## Explore the Golden Ratio

```python
import math

# Golden ratio
golden_ratio = (1 + math.sqrt(5)) / 2

# Check for flags approximating the golden ratio
golden_approximations = [ratio for ratio in unique_ratios if abs(ratio - golden_ratio) < 0.1]

# Print flags approximating the golden ratio
if golden_approximations:
    print(f"Flags approximating the golden ratio: {golden_approximations}")
else:
    print("No flags approximate the golden ratio closely.")
```

## Visualize Flag Design Elements

```python
from PIL import Image
import colorsys

# Function to get dominant color
def get_dominant_color(img_path):
    img = Image.open(img_path)
    img = img.convert('RGB')
    img = img.resize((150, 150))  # Resize for faster processing
    pixels = list(img.getdata())
    r, g, b = 0, 0, 0
    count = 0
    for pixel in pixels:
        r += pixel[0]
        g += pixel[1]
        b += pixel[2]
        count += 1
    r, g, b = r / count, g / count, b / count
    return (r, g, b)

# Get dominant colors for each flag
dominant_colors = {}
for filename in os.listdir(flags_folder):
    if filename.endswith(".png"):
        img_path = os.path.join(flags_folder, filename)
        dominant_colors[filename] = get_dominant_color(img_path)

# Print dominant colors
for filename, color in dominant_colors.items():
    print(f"{filename}: {color}")
```

##  Normalize Flags for Uniform Representation

```python
# Function to normalize a flag to a uniform aspect ratio
def normalize_flag(img_path, target_width=100, target_height=100):
    img = Image.open(img_path)
    width, height = img.size
    aspect_ratio = width / height
    
    # Calculate new dimensions while maintaining aspect ratio
    if width > height:
        new_width = target_width
        new_height = int(target_width / aspect_ratio)
    else:
        new_height = target_height
        new_width = int(target_height * aspect_ratio)
    
    # Resize the image
    img = img.resize((new_width, new_height))
    
    # Create a new image with the target size and paste the resized flag
    new_img = Image.new('RGB', (target_width, target_height), (255, 255, 255))  # White background
    x_offset = (target_width - new_width) // 2
    y_offset = (target_height - new_height) // 2
    new_img.paste(img, (x_offset, y_offset))
    
    return new_img

# Normalize flags
normalized_flags = {}
for filename in os.listdir(flags_folder):
    if filename.endswith(".png"):
        img_path = os.path.join(flags_folder, filename)
        normalized_flags[filename] = normalize_flag(img_path)

# Save normalized flags
for filename, img in normalized_flags.items():
    img.save(f"normalized_{filename}")
```

## New Observations

```python
# Example: Clustering colors
from sklearn.cluster import KMeans
import numpy as np

# Convert dominant colors to numpy array
colors_array = np.array(list(dominant_colors.values()))

# Perform K-means clustering on colors
kmeans = KMeans(n_clusters=5)  # Adjust clusters as needed
kmeans.fit(colors_array)

# Print cluster labels for each flag
for filename, color in dominant_colors.items():
    label = kmeans.predict([color])[0]
    print(f"{filename}: Cluster {label}")
```

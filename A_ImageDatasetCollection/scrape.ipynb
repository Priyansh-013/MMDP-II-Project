```python
import os
import requests
import csv

# Define categories and create folders
categories = [
    "Apple",
    "Banana",
    "Mango",
    "Orange",
    "Grapes",
    "Pineapple",
    "Watermelon",
    "Papaya",
    "Pear",
    "Pomegranate",
    "Peach",
    "Plum",
    "Kiwi",
    "Lemon",
    "Strawberry",
    "Cherry",
    "Dragon Fruit",
    "Custard Apple",
    "Jackfruit",
    "Tamarind"
]


```

```python
import os
import csv
import re
from PIL import Image
import sys
from bing_image_downloader import downloader

# Function to download images for a specific category
def download_images(category, num_images):
    # Create output directory for the category
    output_dir = f'dataset/{category} fruit'
    os.makedirs(output_dir, exist_ok=True)

    # Log file to track downloaded URLs
    log_file = os.path.join(output_dir, "download_log.txt")
    with open(log_file, "w") as log:
        # Redirect stdout to log file temporarily
        original_stdout = sys.stdout
        sys.stdout = log

        topic = f"{category} fruit"
        # Use Bing Image Downloader to fetch images
        downloader.download(topic, limit=num_images, output_dir='dataset',
                            adult_filter_off=True, force_replace=False, timeout=60)

        # Restore stdout after logging
        sys.stdout = original_stdout

    return log_file, output_dir

# Function to extract image URLs from the log file
def extract_urls(log_file):
    urls = []
    with open(log_file, "r") as log:
        for line in log:
            # Match lines containing URLs using regex
            match = re.search(r"Downloading Image #\d+ from (.+)", line)
            if match:
                urls.append(match.group(1))
    return urls
```

```python


# Create metadata CSV file and write headers
metadata_file = 'metadata.csv'
with open(metadata_file, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['Category', 'URL', 'Filename', 'Resolution'])
    

    # Process each category and fetch images
    for category in categories:
        log_file, output_dir = download_images(category, 50)
        
        # Extract image URLs from the log file
        image_urls = extract_urls(log_file)

        # Iterate through downloaded files in the category folder
        for index, filename in enumerate(os.listdir(output_dir)):
            file_path = os.path.join(output_dir, filename)

            try:
                # Open image to retrieve resolution details
                with Image.open(file_path) as img:
                    resolution = f"{img.width}x{img.height}"

                # Get corresponding URL from extracted URLs list
                image_url = image_urls[index] if index < len(image_urls) else "URL Not Available"

                # Write metadata entry into CSV file
                writer.writerow([category, image_url, filename, resolution])

            except Exception as e:
                print(f"Error processing {file_path}: {e}")

print(f"Metadata file '{metadata_file}' created successfully.")
```

```python
import os

# Define the path to your dataset folder
dataset_folder = 'dataset'  # Replace with the actual path

# Check if the dataset folder exists
if not os.path.exists(dataset_folder):
    print("The dataset folder does not exist.")
else:
    # List all folders in the dataset
    categories = [folder for folder in os.listdir(dataset_folder) if os.path.isdir(os.path.join(dataset_folder, folder))]

    for category in categories:
        category_path = os.path.join(dataset_folder, category)
        
        # List all files (images) in the category folder
        images = [img for img in os.listdir(category_path) if os.path.isfile(os.path.join(category_path, img))]
        
        for image in images:
            # Construct the new filename by prefixing with the category
            new_filename = f"{category}_{image}"
            
            # Rename the image
            os.rename(os.path.join(category_path, image), os.path.join(category_path, new_filename))
            
            print(f"Renamed {image} to {new_filename} in {category} folder.")
```

```python
import os

def rename_files_recursively(folder_path):
    for root, dirs, files in os.walk(folder_path):
        folder_name = os.path.basename(root)
        
        for file in files:
            # Check if it's an image file
            if file.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff')):
                file_path = os.path.join(root, file)
                
                # Remove folder name from file name if present
                if file.startswith(folder_name):
                    new_filename = file[len(folder_name):]
                    # Remove any leading underscores or spaces
                    new_filename = new_filename.lstrip('_').lstrip()
                    
                    # Construct new file path
                    new_file_path = os.path.join(root, new_filename)
                    
                    try:
                        os.rename(file_path, new_file_path)
                        print(f"Renamed {file} to {new_filename}")
                    except Exception as e:
                        print(f"Error renaming {file}: {e}")

# Specify the top-level folder path
folder_path = 'dataset'
rename_files_recursively(folder_path)
```

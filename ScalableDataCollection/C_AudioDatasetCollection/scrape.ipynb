```python
import requests
import subprocess
import os
import json
from datetime import datetime

def record_audio(url, station_name, duration, output_format="mp3"):
    """
    Records audio from a given URL for a specified duration and saves it in the specified format.

    Args:
        url (str): The URL of the audio stream.
        station_name (str): Name of the radio station.
        duration (int): Duration to record in seconds.
        output_format (str): Format to save the audio file (e.g., "mp3" or "wav").

    Returns:
        str: Path to the saved audio file.
    """
    file_name = f"{station_name}_{datetime.now().strftime('%Y%m%d%H%M%S')}.{output_format}"
    
    # Use ffmpeg to record the stream for the specified duration
    command = f"ffmpeg -i {url} -t {duration} {file_name}"
    subprocess.run(command, shell=True)
    
    # Save metadata
    metadata = {
        "station_name": station_name,
        "timestamp": datetime.now().strftime('%Y%m%d%H%M%S'),
        "duration": duration,
        "file_path": file_name
    }
    
    with open(f"{station_name}_{datetime.now().strftime('%Y%m%d%H%M%S')}.json", "w") as metadata_file:
        json.dump(metadata, metadata_file)
    
    return file_name

# Example usage
stations = [
    {"url": "http://stream.live.vc.bbcmedia.co.uk/bbc_radio_fourlw_online_nonuk", "name": "BBC Radio 4"},
    # Add more stations here
]

for station in stations:
    for _ in range(30):  # Record 30 files per station
        recorded_file = record_audio(station["url"], station["name"], duration=30, output_format="mp3")
        print(f"Audio recorded and saved as {recorded_file}")
```

```python
import os
from datetime import datetime
import subprocess
import json

def create_dataset_folder():
    """
    Creates a folder named 'dataset' if it does not already exist.
    """
    folder_name = "dataset"
    if not os.path.exists(folder_name):
        os.makedirs(folder_name)
    return folder_name

def record_audio(url, station_name, duration, output_format="mp3"):
    """
    Records audio from a given URL for a specified duration and saves it in the specified format.

    Args:
        url (str): The URL of the audio stream.
        station_name (str): Name of the radio station.
        duration (int): Duration to record in seconds.
        output_format (str): Format to save the audio file (e.g., "mp3" or "wav").

    Returns:
        str: Path to the saved audio file.
    """
    folder_name = create_dataset_folder()
    timestamp = datetime.now().strftime('%Y%m%d%H%M%S')
    file_name = f"{folder_name}/{station_name}_{timestamp}.{output_format}"
    
    # Check if ffmpeg is installed
    try:
        subprocess.run(["ffmpeg", "-version"], check=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    except FileNotFoundError:
        print("ffmpeg is not installed. Please install it first.")
        return None

    # Use ffmpeg to record the stream for the specified duration
    command = f"ffmpeg -i {url} -t {duration} -c:a libmp3lame {file_name}"
    try:
        subprocess.run(command, shell=True, check=True)
    except subprocess.CalledProcessError as e:
        print(f"Error recording audio: {e}")
        return None

    # Save metadata
    metadata = {
        "station_name": station_name,
        "timestamp": timestamp,
        "duration": duration,
        "file_path": file_name
    }

    metadata_file_name = f"{folder_name}/{station_name}_{timestamp}.json"
    with open(metadata_file_name, "w") as metadata_file:
        json.dump(metadata, metadata_file)

    return file_name

# Example usage
stations = [
    {"url": "http://stream.live.vc.bbcmedia.co.uk/bbc_radio_fourlw_online_nonuk", "name": "BBC Radio 4"},
    # Add more stations here
]

for station in stations:
    for _ in range(30):  # Record 30 files per station
        recorded_file = record_audio(station["url"], station["name"], duration=30, output_format="mp3")
        if recorded_file:
            print(f"Audio recorded and saved as {recorded_file}")
```
